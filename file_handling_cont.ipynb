- Deserialization - process of converting JSON in python data types

# What is JSON?
- it is kind of dictionary from python, key-val pairs

```python
# Serialization using JSON module

# list

import json
L = [1,2,3,4,5]

with open('demo.json', 'w') as f:
  json.dump(L, f)

```

```python
# dict
import json
d = {
    'name' : 'karan',
    'age' : 21,
    'gender': 'male'
}

with open('demo.json', 'w') as f:
  json.dump(d, f, indent=4)
```

```python
# Deserialization using JSON module

import json

with open('demo.json', 'r') as f:
  d = json.load(f)
  print(d)
  print(type(d))
```

```python
# tuple (a little different)
import json

t = (1,2,3,4,5)

with open("demo.json", 'w') as f:
  json.dump(t, f)

# this would store a list too! this is a problem.
```

```python
# serialize and deserialize a nested dict
import json

d = {
    'student' : 'karan',
    'marks' : [23,12,14,56,77]
}

with open('demo.json', 'w') as f:
  json.dump(d,f)
```

# Serializing and Deserializing Custom Objects

```python
class Person:

  def __init__(self, fname, lname, age, gender):
    self.fname = fname
    self.lname = lname
    self.age = age
    self.gender = gender

person = Person('Karan', 'Singh', 21, 'male')

import json
with open('demo.json', 'w') as f:
  json.dump(person, f)
```

```python
# alag method hai, python doesn't know how our object looks
class Person:

  def __init__(self, fname, lname, age, gender):
    self.fname = fname
    self.lname = lname
    self.age = age
    self.gender = gender

import json

def show_object(person):
  if isinstance(person, Person):
    return f"{person.fname} {person.lname} age -> {person.age} gender -> {person.gender}"

person = Person("Karan", "Singh", 21, "man")
with open('demo_custom.json', 'w') as f:
  json.dump(person, f, default=show_object)
```

```python
# dict format
class Person:

  def __init__(self, fname, lname, age, gender):
    self.fname = fname
    self.lname = lname
    self.age = age
    self.gender = gender

person = Person("Karan", "Singh", 21, 'man')
import json

def show_object(person):
  return {'name' : person.fname + ' ' + person.lname, 'age': person.age, 'gender': person.gender}

with open('demo.json', 'w') as f:
  json.dump(person, f, default = show_object, indent=4)
```

```python
# deserialize
import json

with open('demo.json', 'r') as f:
  d = json.load(f)
  print(d)
  print(type(d))
```

# Pickling & Unpickling
Pickling is the process whereby a python object hierarchy is converted into a byte stream, and unpickling is the inverse, binary file to object.

```python
class Person:

  def __init__(self, name, age):
    self.name = name
    self.age = age

  def display_info(self):
    print("Hi, My name is", self.name,"and I am", self.age)
```

```python
p = Person('Karan', 21)
```

```python
# pickle dump
import pickle
with open('person.pkl', 'wb') as f:
  pickle.dump(p, f)
```

```python
# pickle load
import pickle
with open('person.pkl', 'rb') as f:
  p = pickle.load(f)

p.display_info()
```

# Pickling vs JSON
pickle lets the user to store data in binary format, JSON is used to store data in a human readable format.
