# Exceptions
it things go wrong during the execution of the program (runtime). It generally happens when something unforeseen has happened.
- Exceptions are raised by python runtime
- You have to tackle this on the fly

## Examples
- Memory Overflow
- Divide by 0 -> logicalError
- Database Error

```python
# Handling Exceptions
# Try Except Block
```

```python
# making a file
with open('sample.txt', 'w') as f:
  f.write('Hello World')
```

```python
# try catch demo, maybe what if the file didn't exist
try:
  with open('sample.txt', 'r') as f:
    print(f.read())
except:
  print("File not Found!")
```

```python
# There's another catch we gotta look out for.
try:
  f = open('sample.txt',  'r')
  print(f.read())
  print(m)  # this still throws an error, to which our code executes the except block.

except:
  print("Something went Wrong!")
```

```python
# So you could avoid this by handling the Error Specifics
try:
  f = open('sample.txt', 'r')
  print(f.read())
  print(m)

except Exception as e:
  print(e)
  print(e.with_traceback)
```

```python
# will write two except blocks for the two potential breaking points
try:
  m = 5

  f = open('sample1.txt', 'r')
  print(f.read())
  print(m)
  print(5/2)

  L = [1,2]
  L[100]

# except FileNotFoundError:
#   print("File not Found!")
except NameError:
  print("Variable not defined")
except ZeroDivisionError:
  print("can't divide by 0")

# last mein ek generic except bhi daal do
except Exception as e:
  print(e)
```

```python
# else block, if try succeeded without an error, we go to else
try:
  f = open('sample1.txt', 'r')

except FileNotFoundError:
  print("File nahi mili")
except Exception as e:
  print('kuch toh lafda hai')

else:
  print(f.read())
```

```python
# finally
# try -> except -> else -> finally

# finally execute no matter what
try:
  f = open('sample.txt', 'r')

except FileNotFoundError:
  print("File nahi mili")
except Exception as e:
  print('kuch toh lafda hai')

else:
  print(f.read())

finally:
  print("ye toh print hoga hee.")
```

```python
class Bank:

  def __init__(self, balance,):
    self.balance = balance

  def withdraw(self, amount):
    if amount < 0:
      raise Exception("Amount cannot be negative")
    if self.balance < amount:
      raise Exception("Insufficient Funds")
    self.balance = self.balance - amount

obj = Bank(10000)

try:
  obj.withdraw(6000)
except Exception as e:
  print(e)
else:
  print(obj.balance)
```

```python
# Creating Custom Exceptions. We do this through our classes

# we can raise it only if it inherits from the Exception class
class MyException(Exception):
  def __init__(self, message):
    print(message)

class Bank:

  def __init__(self, balance,):
    self.balance = balance

  def withdraw(self, amount):
    if amount < 0:
      raise MyException("Amount cannot be negative")
    if self.balance < amount:
      raise MyException("Insufficient Funds")
    self.balance = self.balance - amount

obj = Bank(10000)

try:
  obj.withdraw(56000)
except MyException:
  pass
else:
  print(obj.balance)
```

```python
class SecurityError(Exception):
  def __init__(self, message):
    print(message)

  def logout(self):
    print('Logout')

class Google:
  def __init__(self, name, email, password, device):
    self.name = name
    self.email = email
    self.password = password
    self.device = device

  def login(self, email, password, device):
    if device != self.device:
      raise SecurityError('Bhai teri toh lag gyi')
    if email == self.email and password == self.password:
      print('welcome')
    else:
      print('Login Error')

obj =  Google('Karan', 'karan@gmail.com', 'Karan1234', 'macOS')
try:
  obj.login('karan@gmail.com', 'Karan1234', 'macOS')
except SecurityError as e:
  e.logout()
else:
  # No Exceptions found
  print(obj.name)
finally:
  print('database connection closed')
```
